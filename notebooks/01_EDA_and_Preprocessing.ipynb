# 01 — EDA & Preprocessing
Exploratory data analysis on the real Crop Recommendation dataset and an overview of the synthetic datasets used elsewhere in this project.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

crop_df = pd.read_csv('../dataset/Crop_recommendation.csv')
crop_df.head()
```

## Shape, dtypes, and missing values

```python
print(crop_df.shape)
print(crop_df.dtypes)
print('Missing values:\n', crop_df.isnull().sum())
```

## Class balance

```python
crop_df['label'].value_counts().plot(kind='bar', figsize=(14, 4), color='#2E7D32')
plt.title('Records per Crop')
plt.tight_layout()
plt.show()
```

## Feature distributions

```python
num_cols = ['N', 'P', 'K', 'temperature', 'humidity', 'ph', 'rainfall']
crop_df[num_cols].hist(figsize=(14, 8), bins=30, color='#2E7D32')
plt.tight_layout()
plt.show()
```

## Correlation heatmap

```python
plt.figure(figsize=(8, 6))
sns.heatmap(crop_df[num_cols].corr(), annot=True, cmap='Greens')
plt.title('Feature Correlation')
plt.show()
```

## Synthetic datasets overview
Quick sanity checks on the four synthetic datasets generated by `utils/generate_datasets.py`.

```python
disease_df = pd.read_csv('../dataset/disease_risk.csv')
fert_df = pd.read_csv('../dataset/fertilizer_recommendation.csv')
irrigation_df = pd.read_csv('../dataset/irrigation.csv')
farmer_df = pd.read_csv('../dataset/farmer_records.csv')

for name, df in [('disease_risk', disease_df), ('fertilizer', fert_df),
                 ('irrigation', irrigation_df), ('farmer_records', farmer_df)]:
    print(name, df.shape, 'nulls:', df.isnull().sum().sum())
```

```python
disease_df['risk_level'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(5, 5))
plt.title('Disease Risk Level Balance')
plt.ylabel('')
plt.show()
```
