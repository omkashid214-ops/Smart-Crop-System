# 03 — Disease Risk, Fertilizer & Irrigation Models
Trains the three supporting RandomForest classifiers on the synthetic datasets. Mirrors `utils/train_models.py`.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score, confusion_matrix

RANDOM_STATE = 42
```

## Disease Risk model

```python
disease_df = pd.read_csv('../dataset/disease_risk.csv')
crop_le = LabelEncoder()
disease_df['crop_enc'] = crop_le.fit_transform(disease_df['crop'])
features = ['temperature', 'humidity', 'rainfall', 'leaf_wetness_hours', 'crop_enc']
risk_le = LabelEncoder()
y = risk_le.fit_transform(disease_df['risk_level'])
X = disease_df[features]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y)
disease_model = RandomForestClassifier(n_estimators=250, random_state=RANDOM_STATE, n_jobs=-1)
disease_model.fit(X_train, y_train)
preds = disease_model.predict(X_test)
print('Accuracy:', accuracy_score(y_test, preds))
print('F1 macro:', f1_score(y_test, preds, average='macro'))
```

## Fertilizer Recommendation model

```python
fert_df = pd.read_csv('../dataset/fertilizer_recommendation.csv')
crop_le2 = LabelEncoder()
fert_df['crop_enc'] = crop_le2.fit_transform(fert_df['crop'])
features2 = ['N', 'P', 'K', 'ph', 'crop_enc']
fert_le = LabelEncoder()
y2 = fert_le.fit_transform(fert_df['recommended_fertilizer'])
X2 = fert_df[features2]

X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2, random_state=RANDOM_STATE, stratify=y2)
fert_model = RandomForestClassifier(n_estimators=250, random_state=RANDOM_STATE, n_jobs=-1)
fert_model.fit(X2_train, y2_train)
preds2 = fert_model.predict(X2_test)
print('Accuracy:', accuracy_score(y2_test, preds2))
print('F1 macro:', f1_score(y2_test, preds2, average='macro'))
```

## Irrigation Need model

```python
irrigation_df = pd.read_csv('../dataset/irrigation.csv')
crop_le3 = LabelEncoder()
irrigation_df['crop_enc'] = crop_le3.fit_transform(irrigation_df['crop'])
season_le = LabelEncoder()
irrigation_df['season_enc'] = season_le.fit_transform(irrigation_df['season'])
features3 = ['soil_moisture_pct', 'temperature', 'rainfall_forecast_mm', 'crop_enc', 'season_enc']
need_le = LabelEncoder()
y3 = need_le.fit_transform(irrigation_df['irrigation_needed'])
X3 = irrigation_df[features3]

X3_train, X3_test, y3_train, y3_test = train_test_split(X3, y3, test_size=0.2, random_state=RANDOM_STATE, stratify=y3)
irrigation_model = RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE, n_jobs=-1)
irrigation_model.fit(X3_train, y3_train)
preds3 = irrigation_model.predict(X3_test)
print('Accuracy:', accuracy_score(y3_test, preds3))
print('F1 macro:', f1_score(y3_test, preds3, average='macro'))
```
